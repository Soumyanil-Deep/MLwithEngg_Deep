# Problem 3

```python
W = np.array([[1, -1], [2, 0]])
print(W)
W_size = np.shape(W)

norm1 = 0 # initial
sum_j = 0

for i in range(W_size[1]):
  sum_j = 0 # Reset sum_j for each column
  for j in range(W_size[0]):
    sum_j += abs(W[j][i]) # Use absolute values for L1 norm
  norm1 = max(norm1, sum_j)

print("L1 Norm (Max Abs Column Sum) of W:", norm1)

norm_frob = np.sqrt(np.trace(np.matmul(W.T, W)))

print("Frobenius Norm of W:", norm_frob)

# Calculate L2 norm (spectral norm) - largest singular value
l2_norm_W = np.linalg.svd(W)[1][0]
print("L2 Norm (Spectral Norm) of W:", l2_norm_W)

# Calculate L-infinity norm (maximum absolute row sum norm)
linf_norm_W = np.max(np.sum(np.abs(W), axis=1))
print("L-infinity Norm (Max Abs Row Sum) of W:", linf_norm_W)
```

# Problem 4

```python
from sklearn import datasets

iris = datasets.load_iris()
iris.target_names

X = iris.data[:, :3] # only selecting first three features
y = iris.target

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test=train_test_split(
    X,y,
    test_size=0.40,
    train_size=0.60,
    random_state=123,
    shuffle=True,
    stratify=y)

from sklearn import svm

clf = svm.SVC()
clf.fit(X_train, y_train)

from sklearn.metrics import accuracy_score
acc = accuracy_score(y_test,clf.predict(X_test) )
print('accuracy score :', acc)
```

# Problem 5

```python
import tensorflow as tf
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense
from keras import optimizers
from tensorflow.python.keras.optimizers import *
import numpy as np


#Load dataset
#split into input (X) and output (y)
X = np.arange(1, 101)
X = X*1.0
y = np.arange(1, 101) ** (1 / 7)
y = y*1.0

#define keras model
model = Sequential()

model.add(Dense(6,input_dim=1,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(1))

#compile the keras model
opt = optimizers.Adam(learning_rate=0.001)
mse = tf.keras.losses.MeanSquaredError(
    reduction=tf.keras.losses.Reduction.SUM)
model.compile(loss=mse, optimizer=opt)



#fit the keras model on the dataset (CPU)
model.fit(X,y,epochs=2000,batch_size=10, verbose=0)
model.summary()

#make class predictions with the model
predictions = model.predict(X)

#summarize the first 10 cases
for i in range(10):
    print('%s => %.2f (expected %.2f)' %(X[i].tolist(), predictions[i], y[i]) )
```

# Problem 6

```python
from sklearn import datasets
import numpy as np
from keras.models import Sequential
from keras.layers import Dense
from keras import optimizers
from tensorflow.python.keras.optimizers import *
import tensorflow as tf

# Load the iris dataset
iris = datasets.load_iris()

# Define X (sepal length) and y (petal length)
X_sepal_length = iris.data[:, 0] # Sepal Length is the first feature (index 0)
Y_petal_length = iris.data[:, 2] # Petal Length is the third feature (index 2)

# Reshape X and Y for Keras input (if they are 1D arrays)
X_sepal_length = X_sepal_length.reshape(-1, 1)
Y_petal_length = Y_petal_length.reshape(-1, 1)
```

```python
# Define Keras model
model_iris = Sequential()

model_iris.add(Dense(10, input_dim=1, activation='relu')) # Input is 1 feature (sepal length)
model_iris.add(Dense(10, activation='relu'))
model_iris.add(Dense(1)) # Output is 1 feature (petal length)

# Compile the Keras model
opt_iris = optimizers.Adam(learning_rate=0.001)
mse_iris = tf.keras.losses.MeanSquaredError()
model_iris.compile(loss=mse_iris, optimizer=opt_iris)
```

```python
# Fit the Keras model on the dataset
model_iris.fit(X_sepal_length, Y_petal_length, epochs=200, batch_size=5, verbose=0)
model_iris.summary()

# Make predictions with the model
predictions_iris = model_iris.predict(X_sepal_length)

# Summarize the first 10 cases
print('Predictions vs. Actuals for first 10 samples:')
for i in range(10):
    print(f'Input (Sepal Length): {X_sepal_length[i][0]:.2f} => Predicted (Petal Length): {predictions_iris[i][0]:.2f} (Actual: {Y_petal_length[i][0]:.2f})')
```
